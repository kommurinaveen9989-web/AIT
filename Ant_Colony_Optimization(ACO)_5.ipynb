```python
r1 = 30
r2 = 70
total = r1 + r2
p1 = (r1 / total) * 100
p2 = (r2 / total) * 100
if p2 > p1:
    print(f"When comparing {int(p2)}% and {int(p1)}%, most ants followed the minimum path, i.e., the {int(p2)}% chance path.")
else:
    print(f"When comparing {int(p1)}% and {int(p2)}%, most ants followed the minimum path, i.e., the {int(p1)}% chance path.")
```

```python
import matplotlib.pyplot as plt
r1 = 30
r2 = 70
total = r1 + r2

p1 = (r1 / total) * 100
p2 = (r2 / total) * 100
if p2 > p1:
    chosen_path = "Path 2 (Minimum Path)"
    result_text = f"When comparing {int(p2)}% and {int(p1)}%, most ants followed the minimum path, i.e., the {int(p2)}% chance path."
else:
    chosen_path = "Path 1 (Minimum Path)"
    result_text = f"When comparing {int(p1)}% and {int(p2)}%, most ants followed the minimum path, i.e., the {int(p1)}% chance path."

print(result_text)
paths = ['Path 1 (Longer)', 'Path 2 (Shorter / Min)']
probabilities = [p1, p2]
bar_colors = ['#ff7675', '#00b894']

plt.figure(figsize=(7, 5))
bars = plt.bar(paths, probabilities, color=bar_colors, width=0.45, edgecolor='black', linewidth=1)
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, height + 2, f"{height:.0f}%",
             ha='center', va='bottom', fontsize=12, fontweight='bold')
plt.title("Ant Colony Path Selection Probability", fontsize=13, fontweight='bold', pad=15)
plt.ylabel("Probability (%)", fontsize=11)
plt.ylim(0, 100)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.figtext(0.5, -0.05, result_text, wrap=True, horizontalalignment='center',
            fontsize=10, fontweight='bold', color='#2d3436')

plt.tight_layout()
plt.show()
```
